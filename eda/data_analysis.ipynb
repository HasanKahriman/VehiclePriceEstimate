```python
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (8, 4.5)
```

```python
# Price distribution
plt.hist(df['Price (TRY)'], bins=40, edgecolor='black')
plt.title('Price Distribution')
plt.xlabel('Price (TRY)')
plt.ylabel('Listings')
plt.show()
```

```python
# Average price by brand
df.groupby('Brand')['Price (TRY)'].mean().sort_values().plot(kind='barh')
plt.title('Average Price by Brand')
plt.xlabel('Average Price (TRY)')
plt.show()
```

```python
# Age vs price
plt.scatter(df['Age'], df['Price (TRY)'], alpha=0.3, s=12)
plt.title('Age vs Price')
plt.xlabel('Age (years)')
plt.ylabel('Price (TRY)')
plt.show()
```

```python
# Mileage vs price
plt.scatter(df['Mileage'], df['Price (TRY)'], alpha=0.3, s=12)
plt.title('Mileage vs Price')
plt.xlabel('Mileage (km)')
plt.ylabel('Price (TRY)')
plt.show()
```

```python
# Price by transmission
df.boxplot(column='Price (TRY)', by='Transmission')
plt.suptitle('')
plt.title('Price by Transmission')
plt.ylabel('Price (TRY)')
plt.show()
```

```python
# Sold within 30 days by fuel type
df.groupby('Fuel Type')['Sold'].mean().mul(100).plot(kind='bar')
plt.title('Sold Within 30 Days by Fuel Type')
plt.ylabel('Sold (%)')
plt.xticks(rotation=0)
plt.show()
```
